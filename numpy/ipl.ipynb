```python
import numpy as np
```

```python
data = np.genfromtxt(
    'deliveries.csv',
    delimiter=',',
    names=True,
    dtype=None,
    encoding='utf-8'
)
```

### Task1: Total Runs per match ###

```python
match_id = data['match_id']
runs = data['total_runs']

unique_ids, inverse = np.unique(match_id, return_inverse=True)

total_runs_per_match = np.bincount(inverse, weights=runs)

result = list(zip(unique_ids, total_runs_per_match))

for mid, total in zip(unique_ids, total_runs_per_match):
    print(mid, int(total))
```

### Task2: Top 5 Batters ###

```python
batters = data['batter']
runs = data['batsman_runs'].astype(int)
unique_batters = np.unique(batters)
total_runs = np.zeros(len(unique_batters), dtype=int)
for i, batter in enumerate(unique_batters):
    total_runs[i] = np.sum(runs[batters == batter])
sorted_indices = np.argsort(total_runs)[::-1]
top_5_batters = unique_batters[sorted_indices][:5]
print("Top 5 run scorers in IPL:")
for i, batter in enumerate(top_5_batters):
    print(f"{i+1}. {batter}: {total_runs[unique_batters == batter][0]} runs")
```

### Task3: Strike Rate of Batters ###

```python
batters = data['batter']
runs = data['batsman_runs'].astype(int)

unique_batters, inverse = np.unique(batters, return_inverse=True)

total_runs = np.bincount(inverse, weights=runs)

balls_faced = np.bincount(inverse)

strike_rate = (total_runs / balls_faced) * 100

for b, sr in zip(unique_batters, strike_rate):
    print(b, round(sr, 2))
```

### Task4: Economy Rate of Bowlers ###

```python
bowler = data['bowler']
runs = data['total_runs'].astype(int)

unique_bowlers, inv = np.unique(bowler, return_inverse=True)

runs_conceded = np.bincount(inv, weights=runs)

balls_bowled = np.bincount(inv)

overs = balls_bowled / 6

economy = runs_conceded / overs

economy = np.where(overs > 0, economy, np.inf)

top10_idx = np.argsort(economy)[:10]

top10_bowlers = unique_bowlers[top10_idx]
top10_economy = economy[top10_idx]

for b, e in zip(top10_bowlers, top10_economy):
    print(b, e)
```

### Task5: Runs per over ###

```python
over = data['over']
batsman_runs = data['batsman_runs'].astype(int)

uniq_overs, inv_overs = np.unique(over, return_inverse=True)
runs_in_over = np.bincount(inv_overs, weights=batsman_runs)

balls_in_over = np.bincount(inv_overs)

avg_runs_per_over = (runs_in_over / balls_in_over) * 6

avg_runs_per_over = np.round(avg_runs_per_over, 2)

avg_runs_list = avg_runs_per_over.tolist()

print("\nTask 5: Average Runs per Over")

print("\nRaw Output List:")
print(avg_runs_list)
```

### Task5: Runs Per Over ###

```python
over = np.array(over)
batsman_runs = np.array(batsman_runs)

total_runs = np.bincount(over, weights=batsman_runs, minlength=20)
counts = np.bincount(over, minlength=20)
total_overs = counts / 6

averages = np.divide(total_runs, total_overs, out=np.zeros(20, dtype=float), where=total_overs != 0)
print(averages)
```

### Task6: Boundary Analysis ###

```python
# Extract columns
runs = data['batsman_runs'].astype(int)
teams = data['batting_team']

# Count 4s
num_fours = np.sum(runs == 4)

# Count 6s
num_sixes = np.sum(runs == 6)

# Team with most boundaries (4s + 6s)
boundary_mask = (runs == 4) | (runs == 6)

boundary_teams = teams[boundary_mask]

unique_teams, inverse = np.unique(boundary_teams, return_inverse=True)
boundary_counts = np.bincount(inverse)

top_team = unique_teams[np.argmax(boundary_counts)]

# Output
print("Total 4s:", num_fours)
print("Total 6s:", num_sixes)
print("Top boundary team:", top_team)
```

### Task7: Death Over Analysis ###

```python
overs = data['over'].astype(int)
runs = data['batsman_runs'].astype(int)
teams = data['batting_team']

# Filter death overs (16–20)
death_mask = (overs >= 16) & (overs <= 20)

death_runs = runs[death_mask]
death_teams = teams[death_mask]

# 1. Total runs in death overs
total_death_runs = np.sum(death_runs)

# 2. Team with highest runs in death overs
unique_teams, inverse = np.unique(death_teams, return_inverse=True)
team_runs = np.bincount(inverse, weights=death_runs)

top_team = unique_teams[np.argmax(team_runs)]
top_runs = np.max(team_runs)

# Output
print("Total runs in death overs:", total_death_runs)
print("Top death overs team:", top_team)
print("Runs scored by top team:", int(top_runs))
```

### Task8: Highest Scoring Match ###

```python
match_ids = data['match_id']
runs = data['total_runs'].astype(int)

unique_matches, inverse = np.unique(match_ids, return_inverse=True)

match_runs = np.bincount(inverse, weights=runs)

max_index = np.argmax(match_runs)

highest_match_id = unique_matches[max_index]
highest_runs = match_runs[max_index]

print((highest_match_id, int(highest_runs)))
```

### Task9: Runs per team per match ###

```python
match_ids = data['match_id']
teams = data['batting_team']
runs = data['total_runs'].astype(int)

combined_key = np.array(
    [str(m) + "_" + str(t) for m, t in zip(match_ids, teams)]
)

unique_keys, inverse = np.unique(combined_key, return_inverse=True)

total_runs_per_group = np.bincount(inverse, weights=runs)

for key, total in zip(unique_keys, total_runs_per_group):
    match_id, team = key.split("_")
    print((int(match_id), team, int(total)))
```

### Task10: Match Winner Approximation ###

```python
match_ids = data['match_id']
teams = data['batting_team']
runs = data['total_runs'].astype(int)

combined_key = np.array(
    [str(m) + "_" + str(t) for m, t in zip(match_ids, teams)]
)

unique_keys, inverse = np.unique(combined_key, return_inverse=True)

team_match_runs = np.bincount(inverse, weights=runs)

match_ids_extracted = np.array([int(k.split("_")[0]) for k in unique_keys])
teams_extracted = np.array([k.split("_")[1] for k in unique_keys])

unique_matches = np.unique(match_ids_extracted)

results = []

for mid in unique_matches:
    mask = match_ids_extracted == mid
    match_teams = teams_extracted[mask]
    match_runs = team_match_runs[mask]

    winner_index = np.argmax(match_runs)
    winner_team = match_teams[winner_index]

    results.append((mid, winner_team))

for r in results:
    print(r)
```

### Task11: Match Scorecard Generation ###

```python
match_ids = data['match_id']
teams = data['batting_team']
runs = data['total_runs'].astype(int)

combined_key = np.array(
    [str(m) + "_" + str(t) for m, t in zip(match_ids, teams)]
)

unique_keys, inverse = np.unique(combined_key, return_inverse=True)
team_runs = np.bincount(inverse, weights=runs)

match_list = np.array([int(k.split("_")[0]) for k in unique_keys])
team_list = np.array([k.split("_")[1] for k in unique_keys])

unique_matches = np.unique(match_list)

for mid in unique_matches:
    print(f"Match {mid}:")

    mask = match_list == mid
    match_teams = team_list[mask]
    match_scores = team_runs[mask]

    for team, score in zip(match_teams, match_scores):
        print(f"  {team}: {int(score)} runs")

    print()
```
