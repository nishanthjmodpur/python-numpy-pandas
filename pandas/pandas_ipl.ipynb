```python
import pandas as pd
```

```python
deliveries_df = pd.read_csv(r"C:\Users\nmod3003\OneDrive - 7-Eleven, Inc\Desktop\Development\7-11apps\python\deliveries.csv")
matches_df = pd.read_csv(r"C:\Users\nmod3003\OneDrive - 7-Eleven, Inc\Desktop\Development\7-11apps\python\matches.csv")
```

## Stage 2 ##

```python
deliveries_df = deliveries_df.dropna(subset=["match_id"])
df = deliveries_df.merge(matches_df, left_on="match_id", right_on="id")
df = df.drop_duplicates()
```

### Stage 4 ###

```python
total_runs_per_match = df.groupby("match_id")["total_runs"].sum().reset_index()
print(total_runs_per_match)
```

### 2. Runs per team per match ###

```python
total_runs_per_team_per_match = df.groupby(["match_id", "batting_team"])["total_runs"].sum().reset_index()
print(total_runs_per_team_per_match)
```

### 3. Top 10 Batters ###

```python
top_10_batters = df.groupby("batter")["batsman_runs"].sum().reset_index()
top_10_batters = top_10_batters.sort_values("batsman_runs", ascending=False).head(10)
print(top_10_batters)
```

### 4. Strike Rate of batters ###

```python
batsman_runs_scored = df.groupby("batter")["batsman_runs"].sum().reset_index()
batsman_balls_faced = df.groupby("batter")["batter"].value_counts()

strike_rates = batsman_runs_scored.merge(batsman_balls_faced, on="batter", how="inner")
strike_rates["strike_rate"] = strike_rates["batsman_runs"]/strike_rates["count"] * 100
sortedstr = strike_rates.sort_values(by="strike_rate", ascending=False)
print(sortedstr)
```
